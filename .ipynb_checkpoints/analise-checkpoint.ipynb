## Problemas encontrados nos dados

**visitas.csv**
- Nomes de colunas fora do padrão (' Canal' com espaço, 'Device' e 'Purchase' com letra maiúscula.
- 150 linhas duplicadas.
- 451 valores ausentes em Canal.
- 254 valores ausentes em Device.
- Canal com escrita inconsistente (ads / ADS / " ads" e o mesmo para e-mail e organic).
- Device com escrirta inconsiste (molile / Mobile, descktop / Descktop).

**Pedidos.csv**
- 16 valores ausentes em price.
- Categoria com maiúscula e espaço inscossistente (Bolsas / bolsas / "Bolsas").
- Preço negativo (-45.90) no pedido 8, a ser comunicado à Maria.
